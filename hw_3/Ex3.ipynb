```python
def main_name(name):
    if '(' in name:
        inside = name.split('(')[1].split(')')[0].strip()
        abbreviation = ['Mr.', 'Mrs.', 'Miss.', 'Master.', 'Dr.', 'Rev.', 'Col.', 'Capt.', 'Mme.', 'Mlle.', 'Ms.', 'Sir.', 'Lady.']
        for x in abbreviation:
            if inside.startswith(x):
                inside = inside[len(x):].strip()
                break
        return inside.split()[0] if inside else ''
    else:
        after_comma = name.split(',')[1].strip()
        abbreviation = ['Mr.', 'Mrs.', 'Miss.', 'Master.', 'Dr.', 'Rev.', 'Col.', 'Capt.', 'Mme.', 'Mlle.', 'Ms.', 'Sir.', 'Lady.']
        for x in abbreviation:
            if after_comma.startswith(x):
                after_comma = after_comma[len(x):].strip()
                break
        return after_comma.split()[0] if after_comma else ''

df['MainName'] = df['Name'].apply(main_name)

male_names = df[df['Sex'] == 'male']['MainName']
top_male = male_names.mode()[0]

female_names = df[df['Sex'] == 'female']['MainName']
top_female = female_names.mode()[0]

print(f"Самое популярное мужское имя: {top_male}")
print(f"Самое популярное женское имя: {top_female}")
```

```python
for pclass in sorted(df['Pclass'].unique()):
    print(f"\nКласс {pclass}:")
    for sex in ['male', 'female']:
        names = df[(df['Pclass'] == pclass) & (df['Sex'] == sex)]['MainName']
        if not names.empty:
            main = names.mode()[0]
            print(f"  {sex}: {main}")
```
