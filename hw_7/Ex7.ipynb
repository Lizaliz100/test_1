```python
base_url = 'https://lifehacker.ru/topics/technology/'
response = requests.get('https://lifehacker.ru/topics/technology/')
soup = BeautifulSoup(response.text, 'lxml')
```

```python
soup.find_all('a', class_='lh-small-article-card__link')
```

```python
raw_items = soup.find_all('a', class_='lh-small-article-card__link')
links = [item.get('href') for item in raw_items]
links
```

```python
parsed_urls = []
```

```python
for page_num in range(1, 11):
    url = f'{base_url}/?page={page_num}'
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'lxml')
    raw_items = soup.find_all('a', class_='lh-small-article-card__link')
    parsed_urls += [item.get('href') for item in raw_items]
```

```python
len(parsed_urls)
```

```python
result = []

for url in tqdm(parsed_urls):
    article = {}

    article_url = f'https://lifehacker.ru{url}'
    response = requests.get(article_url)
    soup = BeautifulSoup(response.text, 'lxml')

    article['title'] = soup.find('h1', class_='article-card__title').text
    article['text'] = '\n'.join(p.text.strip() for p in soup.find('article').find_all('p'))

    result.append(article)
```

```python
data = pd.DataFrame(result)
data.head(10)
```
