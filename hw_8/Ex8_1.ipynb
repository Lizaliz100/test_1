```python
import requests
import pandas as pd
```

```python
api_base_url = 'https://anapioficeandfire.com/api'
```

```python
books_list = requests.get(
    f'{api_base_url}/books'
).json()
```

```python
len(books_list)
```

```python
df_books = pd.DataFrame(books_list)
df_books.head()
```

```python
houses_list = requests.get(
    f'{api_base_url}/houses'
).json()
```

```python
len(houses_list)
```

```python
df_houses = pd.DataFrame(houses_list)
df_houses.head()
```

```python
houses_with_Titles = requests.get(
    f'{api_base_url}/houses',
    params={'hasTitles': 'true'}
).json()
```

```python
len(houses_with_Titles)
```

```python
df_houses_with_Titles = pd.DataFrame(houses_with_Titles)
df_houses_with_Titles.head()
```
