```python
class Chocolate:
  def __init__(self, price, chocolate_additive=None):
    if price <= 0:
      raise ValueError('Цена должна быть больше 0!')
    self.price = float(price)
    if chocolate_additive is None:
      self.chocolate_additive = None
    else:
      self.chocolate_additive = str(chocolate_additive)
  def show_my_chocolate(self):
    if self.chocolate_additive is None:
      print(f'Обычный шоколадный батончик')
    else:
      print(f'Шоколадный батончик, добавка: {self.chocolate_additive}')
  def __str__(self):
    return f'Шоколадный батончик, {self.price} рублей'
#Пример использования:
Snickers = Chocolate(150.0,'карамель')
Snickers.show_my_chocolate()
print(Snickers)
print('*' * 20)
Twix = Chocolate(130, 'печенье')
Twix.show_my_chocolate()
print(Twix)
print('*' * 20)
Mars = Chocolate(115.0)
Mars.show_my_chocolate()
print(Mars)
print('*' * 20)
Nuts = Chocolate(-100.0, 'орехи')
Nuts.show_my_chocolate()
print(Nuts)
```
