np.int8 – диапазон от -128 до 127;
np.int16 – диапазон от -32768 до 32767;
np.int32 – диапазон от -2.1·10⁹ до 2.1·10⁹;
np.int64 – диапазон от -9.2·10¹⁸ до 9.2·10¹⁸;
np.uint8 – диапазон от 0 до 255;
np.uint16 – диапазон от 0 до 65535;
np.uint32 – диапазон от 0 до 4.2·10⁹;
np.uint64 – диапазон от 0 до 1.8·10¹⁹;
1) Столбец price: min - 1.750000e+06; max - 1.330000e+07. Подходит диапазон от -2.1·10⁹ до 2.1·10⁹ (np.int32).
2) Столбец area: min - 1650.000000; max - 16200.000000. Подходит диапазон от -32768 до 32767 (np.int16).
3) Столбец bedrooms: min - 1.000000; max - 6.000000. Подходит диапазон от -128 до 127 (np.int8).
4) Столбец bathrooms: min - 1.000000; max - 4.000000. Подходит диапазон от -128 до 127 (np.int8).
5) Столбец stories: min - 1.000000; max - 4.000000. Подходит диапазон от -128 до 127 (np.int8).
6) Столбец parking: min - 0.000000; max - 3.000000. Подходит диапазон от -128 до 127 (np.int8).

7) Для оптимизации столбца furnishingstatus можно использовать переход к категориальным значениям – так как в столбце используются повторяющиеся однотипные элементы (furnished, unfurnished, semi-furnished).

8) В остальных столбцах существуют только значения 'yes' и 'no', поэтому можно использовать переход к булевым значениям.

```python
df['price'] = df['price'].astype(np.int32)
df['area'] = df['area'].astype(np.int16)
df['bedrooms'] = df['bedrooms'].astype(np.int8)
df['bathrooms'] = df['bathrooms'].astype(np.int8)
df['stories'] = df['stories'].astype(np.int8)
df['parking'] = df['parking'].astype(np.int8)
df['furnishingstatus'] = df['furnishingstatus'].astype('category')
categories = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea']
for x in categories:
    df[x] = df[x].map({'yes': True, 'no': False}).astype('bool')
df.info()
```

До оптимизации потребление памяти составляло 55.5+ KB, а после 9.3 KB. То есть потребление памяти уменьшилось примерно на 46 KB.
